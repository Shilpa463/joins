```python
1. select c.customername, o.orderamount from customers c
   join orders o on c.customerid = o.customerid
   order by o.orderamount desc
   limit 5;

2. select * from customers where customername like 'A%' or
   customername like 'R%' or customername like 'S%';

3. select c.customername, o.orderid, o.orderamount, o.status from
   orders o inner join  customers c on o.customerid = c.customerid
   where o.orderamount > 10000 and o.status = 'pending';

4. select customerid,
   datediff( current_date, joindate) as days_associated
   from customers
   where datediff( current_date, joindate ) > 365;

5. select orderid,
   orderamount,
   (orderamount * 0.18) as gst_18,
   (orderamount * 1.18) as total_amount_after_gst,
   date_add(orderdate, interval 15 day) as expected_collection_date
   from orders;

6. select customerid, orderamount
   from orders
   order by orderamount desc
   limit 10 offset 10;

7. select c.customerid, c.customername, c.joindate,
   datediff(current_date, c.joindate) as day_since_registration
   from customers c ;

8. select e.employeename, m.employeename as manager,
   e.department from employees e left join employees m
   on e.managerid = m.employeeid;

9. select customerid, orderamount
 from orders
 order by orderamount desc
 limit 10 offset 10;

 select c.customerid, c.customername, c.joindate,
 datediff(current_date, c.joindate) as day_since_registration
 from customers c ;
 select e.employeename, m.employeename as manager, e.department
 from employees e left join employees m on e.managerid = m.employeeid;

SELECT
    c.customername AS "Customer Name",
    c.joindate AS "Join Date",
    MIN(o.orderdate) AS "Order Date",
    DATEDIFF(MIN(o.orderdate), c.joindate) AS "Days Taken To Place First Order"
FROM customers c
JOIN orders o
    ON o.customerid = c.customerid
GROUP BY
    c.customerid,
    c.customername,
    c.joindate
HAVING
    DATEDIFF(MIN(o.orderdate), c.joindate) BETWEEN 0 AND 30
ORDER BY
    c.joindate;

10. SELECT
    c.customername AS "Customer Name",
    COALESCE(SUM(o.orderamount), 0) AS "Total Purchase Amount",

    CASE
        WHEN COALESCE(SUM(o.orderamount), 0) > 50000
            THEN 'Premium'
        WHEN COALESCE(SUM(o.orderamount), 0) BETWEEN 20000 AND 50000
            THEN 'Gold'
        ELSE 'Silver'
    END AS "Spending Category",

    CASE
        WHEN c.joindate < '2024-01-01'
            THEN 'Loyal Customer'
        ELSE 'New Customer'
    END AS "Loyalty Status"

FROM customers c
LEFT JOIN orders o
    ON c.customerid = o.customerid

GROUP BY
    c.customerid,
    c.customername,
    c.joindate

ORDER BY
    "Total Purchase Amount" DESC;


11. SELECT
    c.customername AS "Customer Name",
    o.orderid AS "Order ID",
    o.orderamount AS "Order Amount"
FROM customers c
JOIN orders o
    ON c.customerid = o.customerid
WHERE o.orderamount > (
    SELECT AVG(orderamount)
    FROM orders
)
ORDER BY o.orderamount DESC;


12. SELECT
    c.customername AS "Customer Name",
    o.orderid AS "Order ID",
    o.orderamount AS "Order Amount"
FROM customers c
JOIN orders o
    ON c.customerid = o.customerid
ORDER BY o.orderamount DESC
LIMIT 5 OFFSET 1;

```
